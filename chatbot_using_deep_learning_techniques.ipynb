````python
'''
## 🔹 1. **Data Preprocessing Techniques**

* **Label Encoding (`LabelEncoder`)**

  * Converts text responses (labels) into integers so the neural network can work with them.
  * Technique: *Categorical encoding*.

* **Tokenization (`Tokenizer`)**

  * Splits sentences into words and maps each word to an integer index.
  * Technique: *Word-level tokenization*.

* **Padding (`pad_sequences`)**

  * Ensures all input sequences have the same length by adding zeros to shorter ones.
  * Technique: *Sequence padding for fixed-length input*.

---

## 🔹 2. **Deep Learning Techniques**

* **Word Embeddings (`Embedding` layer)**

  * Converts each word index into a dense vector (100 dimensions here).
  * Technique: *Learned embeddings* → captures semantic similarity between words during training.

* **Feedforward Neural Network (`Dense` layers)**

  * After flattening, a fully connected hidden layer (`Dense(64, relu)`) learns patterns in the input data.
  * Output layer (`Dense(..., softmax)`) assigns probabilities over possible responses.
  * Technique: *Multi-layer Perceptron (MLP)*.

* **Flattening (`Flatten` layer)**

  * Converts the 2D embedding matrix (sequence × embedding size) into a 1D vector.
  * Technique: *Dimensionality reduction via flattening* (but loses sequence order info).

* **Optimization (`Adam`)**

  * Adaptive optimizer that adjusts learning rates during training.
  * Technique: *Gradient-based optimization*.

* **Loss Function (`SparseCategoricalCrossentropy`)**

  * Measures how well predicted class probabilities match true labels.
  * Technique: *Cross-entropy loss for classification*.

---

## 🔹 3. **Prediction & Inference Techniques**

* **Softmax Activation**

  * Converts raw outputs into a probability distribution across all possible responses.
  * Technique: *Probabilistic classification*.

* **Argmax (`np.argmax`)**

  * Picks the most probable class (response).
  * Technique: *Decision rule for classification*.

* **Inverse Label Encoding (`inverse_transform`)**

  * Converts predicted integer back into human-readable response text.
  * Technique: *Post-processing decoded output*.

---

## 🔹 4. **Training Setup**

* **Epoch-based Training (`fit` with `epochs=50`)**

  * Runs multiple passes over the dataset to adjust weights.
  * Technique: *Batch gradient descent (mini-batch via Keras)*.

---

✅ So, in summary the main techniques are:

* **NLP preprocessing**: tokenization, padding, label encoding
* **Representation learning**: embeddings
* **Neural architecture**: MLP with embeddings (flatten → dense → softmax)
* **Optimization & loss**: Adam optimizer + categorical cross-entropy
* **Prediction decoding**: argmax + inverse label encoding

---

Would you like me to also compare this with **better techniques** (like LSTMs/Transformers for sequence handling) so you see how this “simple chatbot” differs from modern approaches?
'''
````

````python
from sklearn.preprocessing import LabelEncoder #used to convert text labels into numerical labels.
from tensorflow import keras #to build the neural network, do tokenization, etc.
import numpy as np #for array operations
````

````python
#Preparing the Training Data
train_data = [
    "Hello",
    "How are you?",
    "Good morning",
    "Good evening",
    "Nice to meet you",
    "What's up?",
    "How's your day going?",
    "Greetings!",
    "Good afternoon",
    "How can I assist you?",
    "Pleasure to see you",
    "Is there anything I can help with?"
]#list of user‐side input examples

train_labels = [
    "Hi",
    "I'm fine, how about you?",
    "Good morning to you",
    "Good evening, how can I help you?",
    "Nice to meet you too",
    "Not much, just hanging out",
    "It's going well, thank you",
    "Hello!",
    "Good afternoon to you too",
    "I'm here to assist you",
    "Likewise!",
    "Yes, I have a question"
]#matching responses the chatbot should produce
````

````python
'''They are aligned: the i-th element of train_data corresponds to the i-th of train_labels

Possible limitations:

Very small dataset → model can overfit, or not generalize.
No “unknown” category or fallback for things the bot doesn’t know.
Very simple patterns; no handling of grammar, synonyms, etc.

'''
````

````python
'''The default pad_sequences pads at the front (pre-padding) unless specified otherwise, which can have implications.

The “maxlen” is inferred automatically because they didn’t specify maxlen, so it becomes the length of the longest sequence in train_sequences.'''
````

````python
#Data Preprocessing

# 1. Encode the labels: This step is necessary because deep learning models require numerical inputs.
label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(train_labels)
#The LabelEncoder is fitted on the train_labels to learn the unique labels and assign them numerical values using transform method 




# 2. Tokenize text input (train_data): Tokenizer from TensorFlow to tokenize the text data into unique sequences of integers.
tokenizer = keras.preprocessing.text.Tokenizer()
tokenizer.fit_on_texts(train_data)
train_sequences = tokenizer.texts_to_sequences(train_data)
#The Tokenizer is fitted on the train_data to learn the unique words and assign them integer values. The texts_to_sequences method is used to 
#convert the text data into sequences of integers based on the learned mapping. Technique: *Word-level tokenization*

# 3. Pad the sequences so they all have same length
train_sequences = keras.preprocessing.sequence.pad_sequences(train_sequences)
#Finally, the pad_sequences method is used to ensure all sequences have the same length by padding or truncating them
````

````python
'''Input (sequence of word indices)
        ↓
Embedding Layer → 2D word vectors (seq_len × 100)
        ↓
Flatten → 1D vector
        ↓
Dense Layer (64 neurons, ReLU) → learns patterns
        ↓
Output Dense Layer (len(train_labels), Softmax) → probability of each response
'''
````

````python
'''What Is an Epoch? The number of epochs is a hyperparameter that defines the number times that the learning 
algorithm will work through the entire training dataset.

Explanation of Layers:

Embedding Layer: Converts word indices into 100-dimensional vectors. Captures semantic meaning.

Flatten Layer: Turns 2D matrix into 1D vector. Needed before Dense layers.

Dense Layer: Hidden layer with ReLU activation allows non-linear mapping.

Output Layer: Softmax gives probabilities for each possible response.

Epochs: Number of times the model sees the entire dataset. More epochs → better fit, but risk of overfitting.

Potential Issue:

Flattening loses sequence information. For better NLP understanding, LSTM, GRU, or Transformer layers are preferable.'''
````

````python
#Building and Training the Chatbot Model


#Sequential groups a linear stack of layers into a tf.keras.Model.
model = keras.models.Sequential()
#Sequential() is a Keras class for building a linear stack of layers.
#Data flows from the first layer → second layer → … → output layer.
#For this chatbot, we want a simple model: text input → embedding → flatten → dense layers → output response.

#Adding the Embedding Layer
model.add(
    keras.layers.Embedding(
        len(tokenizer.word_index) + 1,  # Total number of unique words in your training data + 1 (because index 0 is reserved for padding).vocabulary size (+1 for padding)
        100,                             # embedding dimension(Size of each word vector. Each word is represented as a 100-dimensional vector.)
        input_length=train_sequences.shape[1]  # fixed input sequence length
    )
)
'''What happens internally:
Input: [1, 5, 2] → Embedding layer → [[0.2,...0.1], [0.5,...0.3], [0.9,...0.7]] (3 × 100 matrix).
Each row is a vector representing a word.'''

#Flattening the Embedding Output: #Flatten Layer: Turns 2D matrix into 1D vector. Needed before Dense layers
model.add(keras.layers.Flatten())
#Example: 3 words × 100 dimensions → flatten → 300-dim vector.
#Note:
#Flattening loses the order of words, which is okay for a small chatbot but not ideal for real NLP tasks.
#For better performance, consider LSTM or GRU layers to preserve sequence information.

#Adding a Hidden Dense Layer:Dense Layer: Hidden layer with ReLU activation allows non-linear mapping.
model.add(keras.layers.Dense(64, activation='relu'))
#Dense layer = fully connected layer. Every input node connects to every neuron in this layer.
#64 neurons: The layer learns 64 features/patterns from the input.
#activation='relu': ReLU = Rectified Linear Unit, introduces non-linearity so the network can learn complex relationships.
            #Formula: f(x) = max(0, x)
            #Helps prevent vanishing gradient issues.
        
#Adding the Output Layer: Softmax gives probabilities for each possible response.        
model.add(keras.layers.Dense(len(train_labels), activation='softmax'))
#Number of neurons = len(train_labels) → one neuron per response.
#activation='softmax': Converts output to probabilities for each class.
#Softmax ensures the sum of probabilities = 1 → allows selection of the most likely response using argmax.

#Compiling the Model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
#Optimizer: 'adam' → adjusts model weights to minimize loss. Popular for NLP tasks.
#Loss function: sparse_categorical_crossentropy → used for multi-class classification when labels are integers.
#Metrics: 'accuracy' → to monitor the fraction of correct predictions during training.
   #Why sparse categorical?
          #Labels are integers (0,1,2,…) instead of one-hot vectors.
          #Saves memory and works directly with integer labels.

#Training the Model
model.fit(train_sequences, encoded_labels, epochs=50)
#train_sequences: Input sequences (padded integers).
#encoded_labels: Corresponding integer labels for responses.
#epochs=50: Model will iterate over the entire training data 50 times.
     #Training process:
          #Input sequence → Embedding → Flatten → Dense → Output probabilities.
          #Loss (sparse_categorical_crossentropy) is calculated between predicted probability and actual label.
          #Optimizer (adam) updates weights to minimize loss.
          #Repeat for 50 epochs → model learns to map input messages to correct responses.
````
